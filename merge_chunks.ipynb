```python
import os
import json
import math
```

```python
#number of chunks to merge
n=5
```

```python
for filename in os.listdir("Transcripts"):
    if filename.endswith(".json"):
        file_path=os.path.join("Transcripts",filename)# Transcripts/filename
        with open (file_path,"r",encoding="utf-8") as f:
            data=json.load(f)
            old_chunk_count=len(data["chunk"])
            new_chunk_group_size=math.ceil(old_chunk_count/n)#if chunkcount/n is decimal means need one more chunk-- so used cieling(ceil(3.1)=4

            new_chunks=[]
            for i in range(new_chunk_group_size):
                start_idx=i*n
                end_idx = min((i+1)*n,old_chunk_count)

                chunk_group=data["chunk"][start_idx:end_idx]

                new_chunks.append({
                    "number": data["chunk"][0]["number"],
                    "title": chunk_group[0]["title"],
                    "start": chunk_group[0]["start"],
                    "start": chunk_group[-1]["end"],
                    "text": " ".join(c["text"] for c in chunk_group)
                    
                })
            ##save file without double .json
            os.makedirs("newTranscripts",exist_ok=True)
            with open(os.path.join("newTranscripts",filename),"w",encoding="utf-8") as transcript_files:
                json.dump({"chunk":new_chunks,"text":data["text"]},transcript_files,indent=4)
```
